# Flow algoritmi

```python
G6 = horadam_cubes(3,2,2)
```

```python
rezultati = usporedi_flow_algoritme(G6, 30)
```

### Edmonds-Karp

```python
print("Prosjek:", rezultati['edmonds_karp']['prosjek'])
print("Medijan", rezultati['edmonds_karp']['medijan'])
print("Minimum:", rezultati['edmonds_karp']['minimum'])
print("Maksimum:", rezultati['edmonds_karp']['maksimum'])
```

### Shortest Augmenting Path

```python
print("Prosjek:", rezultati['shortest_augmenting_path']['prosjek'])
print("Medijan:", rezultati['shortest_augmenting_path']['medijan'])
print("Minimum:", rezultati['shortest_augmenting_path']['minimum'])
print("Maksimum:", rezultati['shortest_augmenting_path']['maksimum'])
```

### Preflow Push

```python
print("Prosjek:", rezultati['preflow_push']['prosjek'])
print("Medijan:", rezultati['preflow_push']['medijan'])
print("Minimum:", rezultati['preflow_push']['minimum'])
print("Maksimum:", rezultati['preflow_push']['maksimum'])
```
